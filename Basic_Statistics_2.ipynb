```python
#Creating an array with the given 15 values.

import numpy as np
durability = np.array([[1.13],
                      [1.55],
                      [1.43],
                      [0.92],
                      [1.25],
                      [1.36],
                      [1.32],
                      [0.85],
                      [1.07],
                      [1.48],
                      [1.20],
                      [1.33],
                      [1.18],
                      [1.22],
                      [1.29]]
)
durability
```

```python
#Creating a dataframe.

import pandas as pd
df = pd.DataFrame(durability, columns=['Durability'])
df
```

```python
#Calculating 99% confidence interval using sample Standard Deviation

from scipy import stats

Mean = df["Durability"].mean()

sd = df["Durability"].std()

# Calculating the sample size
n = len(df)

# Calculating the t-distribution for a 99% confidence interval
t_critical = stats.t.ppf(0.995, df=n-1)  # Using t-distribution since sample size is small(<30) and standard deviation of population is unknown.

# Calculating the margin of error
margin_of_error = t_critical * (sd / np.sqrt(n))

# Calculating the confidence interval
confidence_interval = (Mean - margin_of_error, Mean + margin_of_error)

print("We are 99% confident that the population mean lies between:", confidence_interval)
```

```python
#We first create a dataframe with the given values. Then calculate the mean, standard variation and sample size.
#Then we calculate the t-critical value. We need to calculate t-distribution since the sample size is less than 30 and the standard deviation is unkown.
#Then we calculate the margin of error and finally the confidence interval.
```

```python
#Calculating 99% confidence interval using known Standard Deviation of 0.2

from scipy import stats

Mean = df["Durability"].mean()

sd = 0.2

# Calculating the sample size
n = len(df)

# Calculating the z-distribution for a 99% confidence interval
z_critical = stats.norm.ppf(0.995)  # Using z-distribution since standard deviation of entire population is known

# Calculating the margin of error
margin_of_error = z_critical * (sd / np.sqrt(n))

# Calculating the confidence interval
confidence_interval = (Mean - margin_of_error, Mean + margin_of_error)

print("We are 99% confident that the population mean lies between:", confidence_interval)
```
