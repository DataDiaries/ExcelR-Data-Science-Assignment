```python
#1. Python program checking whether a number is prime or not

def is_prime(num):

    if num <= 1:
        return False

#Checking to see if the number is divisible only my 1 and itself or not.

    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

num = int(input("Enter a number: "))
if is_prime(num):
    print(f"{num} is a prime number.")
else:
    print(f"{num} is not a prime number.")
```

```python
#2. Product of auto generated random variables

import random

                                                                        # Generating two random numbers between 1 and 20
num1 = random.randint(1, 20)
num2 = random.randint(1, 20)

                                                                        # Calculating the correct answer and asking the user
correct_ans = num1 * num2
user_ans = int(input(f"What is the product of {num1} and {num2}? "))

                                                                        # Checking the answer
if user_ans == correct_ans:
    print("Correct answer")
else:
    print(f"Incorrect answer")
```

```python
#3. Squares of Even/Odd numbers between 100 and 200.
#Selecting Odd numbers.

def print_odd_squares(start, end):

    for num in range(start, end + 1):
        if num % 2 != 0:                                   # Checking for odd numbers. They should not be divisible by 2.
            print(f"The square of {num} is {num**2}")

print_odd_squares(100, 200)
```

```python
#4. Word Counter - program to count the number of words in a text.

def word_count(text):

    text = text.lower()                       # Removing punctuation and convert text to lowercase to remove case-sensitive counting

    for char in '.,!?;:()[]{}\"\\\'\\\\/-':   # Replace punctuation with spaces
        text = text.replace(char, " ")
    words = text.split()                      # Splitting the input text into words
    counts = {}                               # Initializing an empty dictionary to store word counts
    for word in words:
        if word in counts:
            counts[word] += 1                 # Adding 1 to counts every time a word is repeated
        else:
            counts[word] = 1                  # Assign 1 to the new word
    return counts

print(word_count('Ask not what your country can do for you. Ask what you can do for your country.'))
```

```python
#5. Check for Palindrome using is_palindrome function.

def is_palindrome(input_string):                                    # Two variables are initialized as string date types using empty quotes: "reverse_string" to hold the "input_string" in reverse
                                                                    # order and "new_string" to hold the "input_string" minus the spaces between words, if any are found.
    new_string = ""
    reverse_string = ""

    for letter in input_string:                                      # Iterate through each letter of the "input_string"

        if letter != " ":                                            # The if-statement checks if the "letter" is not a space.

                                                                     # If True, add the "letter" to the end of "new_string" and to the front of "reverse_string".
                                                                     # If False (if a space is detected), no action is needed. Exit the if-block.
            new_string = new_string + letter.lower()                 #Also, converting to lowercase letter.
            reverse_string = letter.lower() + reverse_string

    if new_string ==  reverse_string:                                # Check if the new string is same as reverse string.

        return True                                                  # If True, the "input_string" contains a palindrome.

    return False                                                     # Otherwise, return False.


print(is_palindrome("Never Odd or Even"))
```
