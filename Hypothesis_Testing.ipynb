```python
# Hypothesis Testing
# 1. State the Hypothesis Statement:

# H0: There is no significant change in weekly operating costs.
# H0: u = 1000 + 5*600 = Rs. 4000 (according to the theoretical formula)   (u=mean)

# H1: The weekly operating costs are more than the theoretical costs.
# H1: u > 4000                                                             (u=mean)

# 2. Calculate the test statistic:
# We can calculate the test statistic by using the formula (sample_mean - theoretical_mean) / (sample_std / sqrt(sample_size))

# sample mean weekly cost = Rs. 3,050
# theoretical mean weekly cost according to the cost model (W = $1,000 + $5X for X = 600 units)
# standard deviation of the weekly cost = 5*25 = 125
# sample size = 25 restaurants

# 3. Determine the Critical Value:

# From the z-table, we can find the critical value for a one-tailed test at a significance level of 0.05. The critical value is 1.645.

# 4. Make a Decision
# Comparing the test statistic with the critical value:


import math

#Given data
sample_mean = 3050
theoretical_mean = 4000
sample_std = 5*25
sample_size = 25

# Calculate the test statistic (t)
t_statistic = (sample_mean - theoretical_mean) / (sample_std / math.sqrt(sample_size))

# Determine the critical value for one-tailed test at alpha = 0.05
alpha = 0.05
critical_value = 1.645          #from z-table

# Compare test statistic with critical value
reject_null = t_statistic > critical_value

# Output results
(t_statistic, critical_value, reject_null)
```

```python
# Test Statistics (t) : -38.0
# Critical Value : 1.645
# Decision : Do not reject the null hypothesis
```

```python
# Conclusion:

# Based on the test statistic and critical value, we accept the null hypothesis.
# H0 is accepted and H1 is rejected.
# This means that the owners claim stating that the weekly operating costs are higher than the theoretical mean is False.
```
