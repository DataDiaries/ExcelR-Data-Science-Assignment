```python
import pandas as pd
df = pd.read_csv('sales_data_with_discounts.csv')
df
```

```python
#Identifying the numerical columns in the dataset.
import numpy as np

numerical_columns = df.select_dtypes(include=np.number).columns
numerical_columns
```

```python
#Mean, median, mode and SD for Volume

print("Volume:")
print("Mean:", df["Volume"].mean())
print("Median:", df["Volume"].median())
print("Mode:", df["Volume"].mode()[0])
print("Standard Deviation:", np.round(df["Volume"].std(),2))
```

```python
#Mean, median, mode and SD for Average Price

print("Avg Price:")
print("Mean:", df["Avg Price"].mean())
print("Median:", df["Avg Price"].median())
print("Mode:", df["Avg Price"].mode()[0])
print("Standard Deviation:", np.round(df["Avg Price"].std(),2))
```

```python
#Mean, median, mode and SD for Total Sales Value

print("Total Sales Value:")
print("Mean:", df["Total Sales Value"].mean())
print("Median:", df["Total Sales Value"].median())
print("Mode:", df["Total Sales Value"].mode()[0])
print("Standard Deviation:", np.round(df["Total Sales Value"].std(),2))
```

```python
#Mean, median, mode and SD for Discount Rate (%)


print("Discount Rate (%):")
print("Mean:", df["Discount Rate (%)"].mean())
print("Median:", df["Discount Rate (%)"].median())
print("Mode:", df["Discount Rate (%)"].mode()[0])
print("Standard Deviation:", np.round(df["Discount Rate (%)"].std(),2))
```

```python
#Mean, median, mode and SD for Discount Amount

print("Discount Amount:")
print("Mean:", df["Discount Amount"].mean())
print("Median:", df["Discount Amount"].median())
print("Mode:", df["Discount Amount"].mode()[0])
print("Standard Deviation:", np.round(df["Discount Amount"].std(),2))
```

```python
#Mean, median, mode and SD for Net Sales Value


print("Net Sales Value:")
print("Mean:", df["Net Sales Value"].mean())
print("Median:", df["Net Sales Value"].median())
print("Mode:", df["Net Sales Value"].mode()[0])
print("Standard Deviation:", np.round(df["Net Sales Value"].std(),2))
```

```python
#Mean: The average value of the data points in a column.  It is calculated by summing all values and dividing by the number of values.

#Median: The middle value when the data points are ordered.  If there's an even number of data points, the median is the average of the two middle values. It's less sensitive to outliers than the mean.

#Mode: The value that appears most frequently in the dataset.  A dataset can have multiple modes or no mode at all.

#Standard Deviation:  A measure of the amount of variation or dispersion of a set of values. A low standard deviation indicates that the values tend to be close to the mean, while a high standard deviation indicates that the values are spread out over a wider range.
```

```python
#2. Data Visualization
```

```python
#Histograms of all the numerical columns.

import matplotlib.pyplot as plt
for col in numerical_columns:
    plt.hist(df[col])
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.title(f'Histogram of {col}')
    plt.show()
```

```python
#From the above histograms, we observe that all the numerical data is right skewed except the 'Discount Rate (%)' which is left skewed.
#This means that for Volume, Avg Price, Total Sales Value, Discount Amount and Net Sales value the Mean<Median<Mode.
#For Discount Rate (%), the Mode<Median<Mode.
```

```python
#Boxplots of all the numerical columns.

for col in numerical_columns:
    plt.figure(figsize=(8, 6))  # Adjust figure size for better visibility
    plt.boxplot(df[col], vert=False)  # Create horizontal boxplot
    plt.xlabel(col)
    plt.title(f'Boxplot of {col}')
    plt.show()
```

```python
# Calculate IQR and identify outliers for each numerical column

for col in numerical_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    lower_bound
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"IQR for {col}: {IQR}")
    print(f"Outliers for {col}:\n{outliers[col]}")
    print("-" * 10)
```

```python
#From the box plots, we can observe that all the numerical columns have outliers to the right except the Discount Rate % which has outliers to the left.
#The IQR shows the spread of the data. The wider the box, the higher the variability.
```

```python
#3. Standardization of Numerical Variables.

#Using standardization, we transform the original value to lie somewhere between -3 to +3.
#When we use standardization, the mean is 0 and standard deviation is 1.
#After standardization, approximately 99.7% of the data lies between -3sigma to +3sigma.
#Approximately 95.5% of the data lies between -2sigma to +2sigma.
#Approximately 68% of the data lies between -1sigma to +1sigma.
#After standardization, we get the z calculated values with which we can determine what percentage of values lie within a certain range.
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

for col in numerical_columns:                                   #Before Standardization.
    print(f"Distribution of {col} before standardization:")
    print(df[col].describe())
    plt.hist(df[col], density=True, alpha=0.6, color='b')
    xmin, xmax = plt.xlim()
    x = np.linspace(xmin, xmax, 100)
    p = norm.pdf(x, df[col].mean(), df[col].std())
    plt.plot(x, p, 'k', linewidth=2)
    plt.title(f'Distribution of {col} before standardization')
    plt.show()

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df_scaled = df.copy()                                          #Creating a copy of the original df to avoid altering the original df.
numerical_cols_to_scale = ['Volume', 'Avg Price', 'Total Sales Value', 'Discount Rate (%)', 'Discount Amount', 'Net Sales Value']
df_scaled[numerical_cols_to_scale] = scaler.fit_transform(df[numerical_cols_to_scale])

for col in numerical_cols_to_scale:                             # After Standardization.
    print(f"\nDistribution of {col} after standardization:")
    print(df_scaled[col].describe())
    plt.hist(df_scaled[col], density=True, alpha=0.6, color='b')
    xmin, xmax = plt.xlim()
    x = np.linspace(xmin, xmax, 100)
    p = norm.pdf(x, df_scaled[col].mean(), df_scaled[col].std())
    plt.plot(x, p, 'k', linewidth=2)
    plt.title(f'Distribution of {col} after standardization')
    plt.show()
```

```python
# From the before and after standardization graphs, we can observe that there is no change in the data distribution.
# This means, the data is already standardized.
```
