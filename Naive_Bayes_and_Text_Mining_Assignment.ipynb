# Naive Bayes and Text Mining:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.simplefilter('ignore')
import re
import string
import nltk
```

# Data Exploration and Preprocessing

```python
data=pd.read_csv('blogs.csv')
```

```python
data
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
data.info()
```

```python
def clean(x):
    x=x.lower()
    x=re.sub('[\.*?/]','',x)
    x=re.sub('[%s]' %re.escape(string.punctuation),'',x)
    x=re.sub('w*\d\w*','',x)
    x=re.sub('[0-9]','',x)
    x=re.sub('[''""..,]','',x)
    return x
cleaned_data=lambda x:clean(x)
```

```python
data['Data']=data.Data.apply(cleaned_data)
data
```

```python
nltk.download('punkt_tab')
data['Data']=data['Data'].apply(word_tokenize)
data
```

```python
stopword=set(stopwords.words('english'))
data['Data']=data['Data'].apply(lambda tokens:[word for word in tokens if word not in stopword])
data
```

```python
data['Data']=[" ".join(doc) for  doc in data['Data']]
tf=TfidfVectorizer()
tfmatrix=tf.fit_transform(data['Data'])

print('TF_TDF MATRIX:','\n',tfmatrix.toarray())
print('COLUMNS NAMES:','\n',tf.get_feature_names_out())
```

```python
extracted_data=pd.DataFrame(tfmatrix.toarray(),columns=tf.get_feature_names_out())
extracted_data
```

# Naive Bayes Model for Text Classification & Evaluation

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,classification_report
features=tf.fit_transform(data['Data'])
target=data['Labels']
```

```python
x_train,x_test,y_train,y_test=train_test_split(features,target,test_size=0.2,random_state=43)
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
model=MultinomialNB()
```

```python
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
print("Accuracy:",accuracy_score(y_test,y_pred))
```

```python
print("Classification Report",classification_report(y_test,y_pred))
```

# Sentiment Analysis

```python
from textblob import TextBlob
def get_sentiment(text):
    analysis=TextBlob(text)
    if analysis.sentiment.polarity>0:
        return 'Positive'
    elif analysis.sentiment.polarity<0:
        return 'Negitive'
    else:
        return 'Neutral'

data['Sentiment']=data['Data'].apply(get_sentiment)
print(data[['Labels','Sentiment']])
```

```python
sentiment_distribution=data.groupby('Labels')['Sentiment'].value_counts()
print(sentiment_distribution)
```

**Evaluation**

1. The Naive Bayes classifier achieved an overall accuracy of 83.25%. Looking at the classification report, we can observe varying performance across different classes. Some classes like 'rec.motorcycles', 'talk.politics.mideast' and 'comp.sys.ibm.pc.hardware' show very strong performance with high precison, recall and F1-scores, often above 0.90.

However, other classes like 'comp.os.ms-windows.misc' (F1 score 0.71), 'rec.sport.hockey' (F1 score 0.67) and 'talk.religion.misc' (F1 score 0.59) had lower performance. For instance, the low precision 're.sport.hockey'(0.50) with high recall(1.00) indicates that it correctly identifies positive cases but also misclassifies many negative cases as positive.

2. Challenges Encountered could include the complexity and ambiguity of natural language, where a simple approach might struggle to capture semantic nuances. Also, potential class imbalance could lead to skewed performance favoring majority classes.

3. The sentiment analysis shows a predominant 'Positive' sentiment across most blog categories, as seen in the sentiment distribution. For instance, 'alt.atheism', 'comp.graphics', 'comp.os.ms-windows.misc', 'misc.forsale', 'rec.autos', 'rec.motrcycles', 'sci.crypt', 'sci.electronics', 'sci.med', 'sci.med', 'sci.space', 'soc.religion.christian', 'talk.politics.mideast' and 'talk.politics.misc' all have a significantly higher count of 'Positive' posts compared to 'Negative' ones.
Some, categories like 'comp.windows.x' and 'talk.politics.guns' show a small number of 'Neutral' sentiments, indicating posts that neither express positive nor negative opinions.
Implications for the content of the blog posts: The overall positive lean could mean that users tend to express constructive opinions, share positive experiences or the topic themselves might elicit more positive engagement. However, the presence of negative sentiment highlights areas of contention or dissatisfaction within the respective communities. A more granular analysis of the content within these negative posts could provide deeper insights into specific problems or disagreements.
